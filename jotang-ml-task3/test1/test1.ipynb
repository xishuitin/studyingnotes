```python
import numpy as np

x = np.array(2.0)
y = np.array(3.0)

a = x*y
b = x ** 2
c = a + b
f = c ** 2
print("前向结果 f =",f)

df_dc = 2*c
df_da = df_dc * 1
df_db = df_dc * 1

df_dy = df_da * x
df_dx = df_da * y + df_db * (2 * x)

print("numpy手动梯度:")
print(f"df/dx = {df_dx:.2f}")
print(f"df/dy = {df_dy:.2f}")
```

```python
import torch

x_t = torch.tensor(2.0, requires_grad=True)
y_t = torch.tensor(3.0, requires_grad=True)

a_t = x_t * y_t
b_t = x_t ** 2
c_t = a_t + b_t
f_t = c_t ** 2

f_t.backward()
print("\nPytorch autograd梯度：")
print(f"df/dx = {x_t.grad:.2f}")
print(f"df/dy = {y_t.grad:.2f}")
```
