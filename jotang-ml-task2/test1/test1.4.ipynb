```python
import numpy as np
from PIL import Image
import matplotlib.pyplot as plt
img = Image.open("outer1.png").convert("L")
img_np = np.array(img, dtype=np.float32)
sobel_y = np.array([
    [-1, -2, -1],
    [ 0,  0,  0],
    [ 1,  2,  1]
], dtype=np.float32)

def conv2d(image,kernel,stride=1,padding="same"):
        img_h, img_w = image.shape
        k_h, k_w = kernel.shape

        if padding == "same":
                pad_h = (k_h - 1) // 2
                pad_w = (k_w - 1) // 2
        else:
                pad_h = 0
                pad_w = 0

        img_pad = np.pad(image, ((pad_h, pad_h), (pad_w, pad_w)), mode='constant')

        out_h = (img_h + 2 * pad_h - k_h) // stride + 1
        out_w = (img_w + 2 * pad_w - k_w) // stride + 1
        out = np.zeros((out_h, out_w), dtype=np.float32)


        for y in range(out_h):
            y_start = y * stride
            for x in range(out_w):
                x_start = x * stride
                window = img_pad[y_start:y_start + k_h, x_start:x_start + k_w]
                out[y, x] = np.sum(window * kernel)

        return out


img_valid = conv2d(img_np, sobel_y, stride=1, padding="valid")
print("原图shape：", img_np.shape)
print("valid输出shape：", img_valid.shape)
def normalize(x):
    return (x - x.min())/(x.max()-x.min()) *255

img_valid_show = normalize(img_valid)
plt.figure()
plt.subplot(1,2,1)
plt.title("valid")
plt.imshow(img_valid_show, cmap="gray")
```
