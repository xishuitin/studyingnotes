```python
import numpy as np
from PIL import Image
import matplotlib.pyplot as plt

def conv2d_manual(img, kernel, padding="valid", stride=1):
    k_h, k_w = kernel.shape
    H, W, C = img.shape

    if padding == "same":
        pad_h = (k_h - 1) // 2
        pad_w = (k_w - 1) // 2
    else:
        pad_h = 0
        pad_w = 0
    img_pad = np.pad(img, ((pad_h, pad_h), (pad_w, pad_w), (0, 0)), mode='constant')

    out_h = (H + 2 * pad_h - k_h) // stride + 1
    out_w = (W + 2 * pad_w - k_w) // stride + 1
    out = np.zeros((out_h, out_w, C), dtype=np.float32)

    for c in range(C):
        for y in range(out_h):
         for x in range(out_w):
            y_start = y * stride
            x_start = x * stride
            patch = img_pad[y_start:y_start + k_h, x_start:x_start + k_w, c]
            out[y, x, c] = np.sum(patch * kernel)
    return out


```
