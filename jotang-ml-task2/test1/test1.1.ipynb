```python
from PIL import Image
import numpy as np
import matplotlib.pyplot as plt
import torch

img_pil = Image.open("outer1.png").convert("RGB")

print("Pillow image size:", img_pil.size)
print("Pillow image mode:", img_pil.mode)

plt.figure(figsize=(8, 5))
plt.imshow(img_pil)
plt.axis('off')
plt.title("Image")
plt.show()
```

```python
img_np = np.array(img_pil)
print(" ====Numpy HWC ====")
print(f"shape: {img_np.shape}")
print(f"Data type: {img_np.dtype}")
print(f"max:{img_np.max()}, min:{img_np.min()}")

pixel = img_np[100,200]
print(f"坐标（行100，列200）像素RGB: {pixel}")

img_tensor = torch.from_numpy(img_np).permute(2, 0, 1)
print(" ====PyTorch CHW Tensor ====")
print(f"shape: {img_tensor.shape}")
print(f"Data type: {img_tensor.dtype}")
print(f"max:{img_tensor.max()}, min:{img_tensor.min()}")
```
